## 43 Pago automático recurrente

```python
# ======================================================
# 💳 CBTis 128
# 📚 Elabora Proyectos de Programación Lógica
# 🤖 Especialidad: Inteligencia Artificial
# 👨‍🏫 Docente: Eliseo Nava
# 👨‍🎓 Alumno: Marco Morquecho
# 💳 Proyecto: IA de Pago Automático Recurrente
# ======================================================

print("========= SISTEMA IA DE PAGOS 💳 =========")

while True:

    nombre = input("\nIngresa nombre del cliente: ")

    saldo = int(input("Ingresa saldo disponible: "))

    pago = int(input("Ingresa monto del pago: "))

    suscripcion = input("¿Tiene suscripción activa? si/no: ")

    retrasos = int(input("Ingresa cantidad de retrasos: "))

    # ======================================================
    # 🧠 Inteligencia Artificial
    # ======================================================

    if saldo >= pago:

        print("Pago aprobado correctamente")

    else:

        print("Saldo insuficiente")

    if pago >= 5000:

        print("Pago alto detectado")

    elif pago >= 1000:

        print("Pago medio detectado")

    else:

        print("Pago pequeño detectado")

    if suscripcion == "si" and retrasos == 0:

        print("Cliente con pagos seguros")

    elif retrasos >= 1 and retrasos <= 3:

        print("Cliente con algunos retrasos")

    else:

        print("Cliente con riesgo de cancelación")

    # ======================================================
    # 📊 Resultado final
    # ======================================================

    print("\n========= RESULTADOS =========")

    print("Cliente:", nombre)

    print("Saldo:", saldo)

    print("Pago:", pago)

    print("Suscripción activa:", suscripcion)

    print("Retrasos:", retrasos)

    continuar = input("\n¿Deseas analizar otro cliente? si/no: ")

    if continuar == "no":

        print("\nPrograma finalizado 💳")

        break
```
